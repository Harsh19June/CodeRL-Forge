````python
# Cell 2: Load and Quantize the Model (Updated)
from unsloth import FastLanguageModel, PatchFastRL
PatchFastRL("GRPO", FastLanguageModel)

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "Qwen/Qwen2.5-Coder-3B-Instruct",
    max_seq_length = 1024,
    load_in_4bit = True,
    fast_inference = False, # CHANGED: Disabled to bypass the vLLM architecture bug
    max_lora_rank = 16,
)

model = FastLanguageModel.get_peft_model(
    model,
    r = 16,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    lora_alpha = 16,
    use_gradient_checkpointing = "unsloth",
)
````

````python
# Cell 4: Load and Format MBPP Dataset (Updated Schema)
from datasets import load_dataset

# 1. Load the sanitized MBPP dataset (with the correct namespace)
raw_dataset = load_dataset("google-research-datasets/mbpp", "sanitized", split="train")

# 2. Define the prompt format with clear system instructions
SYSTEM_PROMPT = """You are an expert Python programmer.
Solve the following programming task.
First, explain your reasoning inside <think>...</think> tags.
Then, provide the complete, working Python solution inside a single ```python ... ``` block.
Do not use input() or interactive functions. Keep functions self-contained."""

def format_prompt(sample):
    # Changed sample['text'] to sample['prompt'] to match the new dataset schema
    prompt_content = f"{SYSTEM_PROMPT}\n\nTask: {sample['prompt']}\nWrite a function that solves this task."
    
    return {
        "prompt": [
            {"role": "user", "content": prompt_content}
        ],
        "test_list": sample["test_list"],
    }

# 3. Map across the dataset and remove only the columns that actually exist
train_dataset = raw_dataset.map(
    format_prompt, 
    remove_columns=['source_file', 'task_id', 'code', 'test_imports']
)

# Sanity check: inspect the first formatted prompt
print("Dataset size:", len(train_dataset))
print("Sample prompt:", train_dataset[0]["prompt"][0]["content"])
print("Sample tests:", train_dataset[0]["test_list"])
````

````python
# Cell 5: Deterministic Compiler & Execution Verifiers
import re
import multiprocessing

def extract_python_code(completion_text: str) -> str:
    """Extracts python code from the markdown code fence."""
    pattern = r"```python\s*(.*?)\s*```"
    matches = re.findall(pattern, completion_text, re.DOTALL)
    if matches:
        return matches[-1].strip() # Use the final code block
    return ""

def _sandboxed_worker(code_str: str, tests: list, result_queue):
    """Executes code and assertions in an isolated namespace."""
    try:
        # Build an isolated execution scope
        local_scope = {}
        exec(code_str, local_scope)
        
        # Execute each assertion from the test suite
        passed_count = 0
        for test_case in tests:
            exec(test_case, local_scope)
            passed_count += 1
            
        result_queue.put({"success": True, "passed": passed_count, "total": len(tests)})
    except SyntaxError:
        result_queue.put({"success": False, "error": "syntax"})
    except Exception as e:
        result_queue.put({"success": False, "error": "runtime", "msg": str(e)})

# --- REWARD FUNCTION 1: Syntax & Formatting ---
def format_and_syntax_reward(prompts, completions, **kwargs) -> list[float]:
    """Penalizes missing tags or invalid Python syntax."""
    rewards = []
    for completion in completions:
        content = completion[0]["content"] if isinstance(completion, list) else completion
        code = extract_python_code(content)
        
        if not code:
            rewards.append(-2.0) # Failed to output a python markdown block
            continue
            
        try:
            compile(code, "<string>", "exec")
            rewards.append(0.5) # Code is syntactically valid
        except SyntaxError:
            rewards.append(-1.0) # Syntax error in generated code
            
    return rewards

# --- REWARD FUNCTION 2: Deterministic Execution Verifier ---
def execution_unit_test_reward(prompts, completions, test_list, **kwargs) -> list[float]:
    """Runs test assertions against the generated code with a hard timeout."""
    rewards = []
    
    for i, completion in enumerate(completions):
        content = completion[0]["content"] if isinstance(completion, list) else completion
        code = extract_python_code(content)
        tests = test_list[i] if i < len(test_list) else []
        
        if not code or not tests:
            rewards.append(0.0)
            continue
            
        # Spawn an isolated process to run assertions
        result_queue = multiprocessing.Queue()
        process = multiprocessing.Process(
            target=_sandboxed_worker, 
            args=(code, tests, result_queue)
        )
        process.start()
        process.join(timeout=2.0) # 2-second timeout per completion
        
        if process.is_alive():
            process.terminate()
            process.join()
            rewards.append(-1.5) # Penalty for infinite loops / hanging execution
            continue
            
        if not result_queue.empty():
            res = result_queue.get()
            if res.get("success"):
                ratio = res["passed"] / res["total"]
                # Scale reward: +2.0 for all tests passed, partial reward for partial passes
                rewards.append(2.0 * ratio)
            elif res.get("error") == "syntax":
                rewards.append(-0.5)
            else:
                rewards.append(-0.2) # Runtime exception (e.g., IndexError, TypeError)
        else:
            rewards.append(-1.0)
            
    return rewards
````

````python
# Cell 6: Configure and Initialize the GRPO Trainer
from trl import GRPOConfig, GRPOTrainer

# 1. Set the Memory-Constrained Hyperparameters
training_args = GRPOConfig(
    output_dir = "grpo_coder_outputs",
    learning_rate = 5e-6,
    lr_scheduler_type = "cosine",
    logging_steps = 10,
    max_steps = 300, # Train for 300 steps minimum to see reasoning emerge
    per_device_train_batch_size = 1,
    gradient_accumulation_steps = 4,
    
    # CRITICAL VRAM SETTINGS FOR KAGGLE:
    num_generations = 4,         # Generates 4 code samples to compare against each other
    max_prompt_length = 256,     # Limit the size of the MBPP instructions
    max_completion_length = 512, # Enough room for reasoning (<think>) + code
)

# 2. Initialize the Trainer
trainer = GRPOTrainer(
    model = model,
    processing_class = tokenizer,
    # Pass your custom deterministic compiler sandbox here:
    reward_funcs = [format_and_syntax_reward, execution_unit_test_reward], 
    args = training_args,
    train_dataset = train_dataset,
)

print("✅ GRPOTrainer successfully initialized and ready.")
````

````python
# Cell 7: Start the RL Loop
trainer.train()
````

````python
# Cell 8: Save the Fine-Tuned Model
model.save_pretrained_merged("grpo_coder_final", tokenizer, save_method = "lora")
print("✅ Model saved successfully!")
````
